# Bagging, 

short for 

Bootstrap Aggregating, is a powerful ensemble learning technique used in machine learning to improve the stability and accuracy of machine learning algorithms. Here's a breakdown of how it works:

Core Idea:

    Bagging aims to reduce variance, a common problem in complex models that tend to overfit the training data.
    It achieves this by creating multiple versions of a predictor from the same training dataset and then combining their outputs.

Steps Involved:

    Bootstrap Sampling:
        The original training dataset is sampled with replacement multiple times (e.g., hundreds or thousands of times). This means that some data points may be included multiple times in a single sample, while others may be omitted.
        Each of these samples is called a "bootstrap sample," and they are the same size as the original dataset.

    Model Training:
        A base learning algorithm (e.g., decision trees, neural networks) is trained on each bootstrap sample, resulting in multiple models.
        Each model is trained independently of the others.

    Aggregation:
        For classification problems, the final prediction is typically made by taking a majority vote among the predictions of all the individual models.
        For regression problems, the final prediction is usually the average of the predictions from all the models.

```python
import pandas as pd
import numpy  as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.metrics import classification_report
```

```python
url ="https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

names =['preg','plas','pres','skin','test','mass','pedi','age','class']
```

```python
data =pd.read_csv(url, names= names)
```

```python
data.head()
```

```python

#split data into X , Y dataset

dataarray =data.values

X = dataarray[:,0:8]

y =dataarray[:,8]
```

```python
from sklearn.model_selection import KFold,cross_val_score

Kfold = KFold(n_splits=10)

cart =DecisionTreeClassifier()

num_trees =100
```

```python
#create bagging model

model =BaggingClassifier(estimator = cart , n_estimators=num_trees, random_state=12)

result =cross_val_score(model, X,y, cv =Kfold)

print(result.mean())
```

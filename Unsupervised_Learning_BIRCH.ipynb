# BIRCH
In machine learning, BIRCH stands for Balanced Iterative Reducing and Clustering using Hierarchies. It's a hierarchical clustering algorithm designed to handle very large datasets efficiently. Here's a breakdown of its key aspects:

Core Concepts:

    Hierarchical Clustering:
        BIRCH creates a hierarchical structure of clusters, meaning clusters can contain sub-clusters, forming a tree-like representation.
    Scalability:
        Its primary strength is its ability to handle large datasets with limited memory. It achieves this by summarizing data into "Clustering Features" (CFs).
    Clustering Features (CFs):
        A CF is a compact representation of a cluster, storing essential statistics like the number of data points, their linear sum, and their squared sum. This allows the algorithm to work with summaries rather than the entire dataset.
    CF Tree:
        BIRCH constructs a CF tree, a hierarchical data structure that stores the CFs. This tree is built incrementally as data points are processed.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

#applying default setting for seaborn
sns.set()

from sklearn.datasets import make_blobs
from sklearn.cluster import Birch
```

```python
#create a cluster from make_blob

X,clusters =make_blobs(n_samples=450 ,centers=6 ,cluster_std=0.7,random_state=0)

plt.scatter(X[:, 0], X[:,1] , alpha=0.7 ,edgecolors='b')
```

```python
brc = Birch(branching_factor=50 ,n_clusters=None, threshold=1.5)

brc.fit(X)
```

```python
labels = brc.predict(X)
```

```python
#final output using BIRCH

plt.scatter(X[:,0] ,X[:,1] , c=labels ,cmap='rainbow' ,alpha=0.7, edgecolors='b')
```

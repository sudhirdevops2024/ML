Description about Boruta Py

BorutaPy is a Python implementation of the Boruta feature selection method, and it's a valuable tool in machine learning for several reasons. Here's a breakdown of its uses:  

Core Purpose: Feature Selection

    Identifying Relevant Features:
        The primary goal of BorutaPy is to determine which features in a dataset are genuinely important for predicting the target variable.   

It goes beyond simply ranking features; it aims to find "all relevant" features, meaning it tries to capture all the features that carry useful information.  

Reducing Noise and Redundancy:

    Datasets often contain many features, some of which may be irrelevant or redundant. BorutaPy helps to eliminate these, leading to cleaner and more efficient models.   

Improving Model Performance:

    By focusing on the most relevant features, BorutaPy can improve the accuracy and generalization ability of machine learning models.   

        It can also prevent overfitting, which occurs when a model learns the noise in the data rather than the underlying patterns.
    Enhancing Interpretability:
        A model with fewer, more relevant features is easier to understand and interpret. This is particularly important in domains where explainability is crucial.

How BorutaPy Works (in simplified terms):

    It creates "shadow features" by making shuffled copies of the original features.   

It trains a machine learning model (often a Random Forest) on the extended dataset, which includes both the original and shadow features.  
It compares the importance of the original features to the importance of the shadow features.  
Features that consistently outperform the shadow features are deemed important.  

Key Benefits:

    "All-relevant" Feature Selection:
        Unlike some other methods that aim for a minimal set of features, BorutaPy tries to find all features that are relevant to the outcome.   

Robustness:

    The use of shadow features and statistical tests makes BorutaPy a robust method for feature selection.

Compatibility:

    BorutaPy is compatible with various ensemble machine learning models from scikit-learn.
    
 BorutaPy is a powerful tool for data scientists and machine learning practitioners who want to identify the most important features in their datasets. It helps to build more accurate, efficient, and interpretable models.

```python
#import required libraries

import pandas as pd
import numpy  as np

from sklearn.ensemble import  RandomForestClassifier
from boruta import BorutaPy
from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score
```

```python
#import data from github

URL ="https://raw.githubusercontent.com/Aditya1001001/English-Premier-League/master/pos_modelling_data.csv"

df = pd.read_csv(URL)
```

```python

#read first 5 records
df.head()
```

```python

#get dataframe structure
df.info()
```

```python
# as position variable is output drop it from dataset and create Independent / input  X variable dataset

x = df.drop('Position',axis= 1)
```

```python
#create dependent / output variable dataset that we are predicting

y= df['Position']
```

```python
#split data in train, test dataset
x_train,x_test, y_train, y_test  = train_test_split(x,y, test_size=0.2, random_state= 1)
```

```python
#fit random forest classifier
rf_all_features =RandomForestClassifier(random_state= 1 ,n_estimators =1000 ,max_depth =5)
```

```python
rf_all_features
```

```python
#try to fit the model and check the accuracy scope
rf_all_features.fit(x_train , y_train)

accuracy_score(y_test, rf_all_features.predict(x_test))
```

```python
#check indepedent train array

np.array(x_train)
```

```python
#apply baruta for random forest model

rfc =RandomForestClassifier(random_state= 1 ,n_estimators =1000 ,max_depth =5)

boruta_selector =BorutaPy(rfc, n_estimators='auto', verbose=2, random_state= 1)

boruta_selector.fit(np.array(x_train), np.array(y_train))
```

```python
#print ranking generated by boruta selector

print('ranking:' , boruta_selector.ranking_)
print("no. of signifant features " , boruta_selector.n_features_)
```

```python
#get the list of columns available in training set

list(x_train.columns)
```

```python
#create dataframe of feature names and ranking and sort it with raking

select_rf_features =pd.DataFrame({'features':list(x_train.columns) ,'ranking':boruta_selector.ranking_ })
select_rf_features.sort_values(by= 'ranking')
```

```python
#remove below features from train set and its rank is more than 1 
#Penalties saved , Own goals,Red cards

x_filter_train = x_train.drop(['Penalties saved','Own goals','Red cards'], axis= 1)

x_filter_test = x_test.drop (['Penalties saved','Own goals','Red cards'], axis= 1)

x_filter_train.info()
```

```python
x_filter_train.shape
```

```python
x_filter_test.shape
```

```python

#run boruta method to indetify relevant feature

boruta_selector.fit(np.array(x_filter_train), np.array(y_train))
```

```python



#select important feature



#x_imp_train =boruta_selector.transform(np.array(x_train))

x_imp_train = boruta_selector.transform(np.array(x_filter_train))

```

```python
#x_filter_test
#x_imp_test= boruta_selector.transform(np.array(x_test))

x_imp_test= boruta_selector.transform(np.array(x_filter_test))
```

```python
rf_boruta =RandomForestClassifier(random_state=1, n_estimators= 1000, max_depth=5)
```

```python
rf_boruta.fit(x_imp_train, y_train)

```

```python
#checking revised accuracy after removing irrelevant features

accuracy_score(y_test, rf_boruta.predict(x_imp_test))
```
